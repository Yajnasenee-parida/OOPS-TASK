3.  In Python, multiple inheritance is a feature where a class can inherit attributes and methods from more than one parent class.

This allows you to combine functionality from multiple sources into a single class.

```python
# 3

class Father:
    def skills_father(self):
        return "Gardening, Fishing"

class Mother:
    def skills_mother(self):
        return "Cooking, Painting"

class Child(Father, Mother):
    def skills_child(self):
        return "Learning, Playing"

# Demonstration of multiple inheritance
if __name__ == "__main__":
    try:
        child_obj = Child()
        print(f"Child's father's skills: {child_obj.skills_father()}")
        print(f"Child's mother's skills: {child_obj.skills_mother()}")
        print(f"Child's own skills: {child_obj.skills_child()}")
    except Exception as e:
        print(f"An error occurred: {e}")
```

4.   In Python, getter methods are used to retrieve the value of a private or protected attribute, while setter methods are used to update or modify the value of that attribute. They allow you to control how attributes are accessed or changed, enabling validation, transformation, or other logic before assignment or retrieval

```python
# 4

class Person:
    def __init__(self, name):
        self._name = name  # Private attribute

    # Getter method
    def get_name(self):
        return self._name

    # Setter method
    def set_name(self, new_name):
        if isinstance(new_name, str) and len(new_name) > 0:
            self._name = new_name
        else:
            print("Invalid name. Name must be a non-empty string.")

# Demonstration
if __name__ == "__main__":
    try:
        person1 = Person("Alice")
        print(f"Initial name: {person1.get_name()}")

        person1.set_name("Bob")
        print(f"Updated name: {person1.get_name()}")

        person1.set_name(123) # Attempt to set an invalid name
        print(f"Name after invalid attempt: {person1.get_name()}")
    except Exception as e:
        print(f"An error occurred: {e}")
```

5.   Method overriding in Python is an object-oriented programming concept where a child (subclass) provides its own implementation of a method that is already defined in its parent (superclass).

When the method is called on an object of the subclass, Python uses the subclass’s version instead of the parent’s version.
This allows you to customize or extend the behavior of inherited methods.

```python
# 5

class Animal:
    def speak(self):
        return "Animal makes a sound"

class Dog(Animal):
    def speak(self):
        return "Woof! Woof!"

# Demonstration of method overriding
if __name__ == "__main__":
    try:
        animal = Animal()
        dog = Dog()

        print(f"Animal speaks: {animal.speak()}")
        print(f"Dog speaks: {dog.speak()}")
    except Exception as e:
        print(f"An error occurred: {e}")
```
